# A Plant-Based Coffee Shop


Coffee is a global phenomenon. Across the world, thousands of new coffee shops open each year, with specialty shops becoming increasingly popular.

In this project we will analyze a survey dataset of approximately 1000 coffee enthusiasts to identify the most popular dairy choices. Based on your findings, you will recommend which dairy alternatives should be stocked by a new specialty coffee shop planning to offer only plant-based beverages.

### Loading the Data
The `coffee-survey-results.csv` file contains information about people's coffee preferences, including which dairy they prefer. 


```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
survey = pd.read_csv('coffee-survey-results.csv')
survey 
```


### Narrowing down the columns
Let's look through all the columns to identify which ones are needed to answer the question about dairy preferences.

```python
survey.columns
```

Columns numbered 4-11 ask about dairy preferences. We can now extract the columns we need for our analysis.


```python
needed_columns = [
    "What kind of dairy? (Whole milk)",
    "What kind of dairy? (Skim milk)",
    "What kind of dairy? (Half and half)",
    "What kind of dairy? (Coffee creamer)",
    "What kind of dairy? (Flavored creamer)",
    "What kind of dairy? (Oat milk)",
    "What kind of dairy? (Almond milk)",
    "What kind of dairy? (Soy milk)",
]
dairy = survey[needed_columns]
dairy
```

This narrows the data to just the 8 columns about dairy preferences.

### Renaming the columns

We can shorten these column names . 

```python
name_map = {
    'What kind of dairy? (Whole milk)': 'Whole milk',
    'What kind of dairy? (Skim milk)': 'Skim milk',
    'What kind of dairy? (Half and half)': 'Half and half',
    'What kind of dairy? (Coffee creamer)': 'Coffee creamer',
    'What kind of dairy? (Flavored creamer)': 'Flavored creamer',
    'What kind of dairy? (Oat milk)': 'Oat milk',
    'What kind of dairy? (Almond milk)': 'Almond milk',
    'What kind of dairy? (Soy milk)': 'Soy milk',
}
```

In the name map above, the new name for 'What kind of dairy? (Whole milk)' is 'Whole milk'. Let's use this name map to rename the column names and improve readability.


```python
dairy = dairy.rename(columns=name_map)
dairy 
```

These column names are much easier to work with.

### Making sense of the data


Here each row represents one coffee enthusiast and their yes/no answers to each survey question. If the enthusiast adds a particular type of dairy to their coffee, that column will show a value of 1, while a value of 0 indicates that they do not use that type of dairy.

Notice that there are also several `NaN` values (Not a Number), which represents missing values. In this case, NaNs indicate that the survey question was not answered.

Let's use the `isna()` method to take a closer look at how many `NaN` values each column has.


```python
dairy.isna().sum()
```

Notice that each dairy option has the same number of NaN values. Why might this be the case?

This particular survey had a question that asks if you add any dairy to your coffee. If the respondent answered "No", the dairy part of the survey was skipped entirely.

Since these rows have no data, we can drop them using `dropna()`.



```python
dairy = dairy.dropna()
dairy 
```

After dropping all the rows with NaNs, there are now 523 rows left in the dataframe. Many rows were removed, such as the row with index 1.

Now that we have cleaned the data, we can investigate dairy preferences.

### What dairy options are most preferred?

What percentage of people consume each type of dairy? We can use the `mean()` method on the dataframe to calculate the ratio of "yes" answers for each column. Multiplying by 100 will give us these ratios as percentages.


```python
dairy_preferences = dairy.mean() * 100
dairy_preferences
```

It looks like whole milk is the most popular. A quick scan of these numbers tells us that the percentages add up to more than 100%. This happens because people often like more than one type of dairy in their coffee.

### Visualizing dairy preferences
Let's visualize the dairy preferences as a horizontal bar chart.

```python
dairy_preferences = dairy_preferences.sort_values()
dairy_preferences 
```

And now we can create our horizontal bar plot.


```python
plt.figure(figsize=(6, 3))
plt.barh(dairy_preferences.index, dairy_preferences)
plt.xlabel("Percent")
```

The most popular plant based dairy choice is oat milk. This might be a great choice for the default dairy option in a specialty coffee shop.

## Project extensions


### Analyzing Sweetener Preferences

Similar to the dairy analysis, we will now identify the most popular sweetener choices. First, we need to extract the relevant columns related to sweetener preferences from the original `survey` DataFrame.

```python
sweetener_columns = [
    'What kind of sugar or sweetener? (Granulated Sugar)',
    'What kind of sugar or sweetener? (Artificial Sweetener)',
    'What kind of sugar or sweetener? (Honey)',
    'What kind of sugar or sweetener? (Maple Syrup)',
    'What kind of sugar or sweetener? (Stevia)',
    'What kind of sugar or sweetener? (Agave Nectar)',
    'What kind of sugar or sweetener? (Brown Sugar)',
    'What kind of sugar or sweetener? (Raw Sugar)',
]
sweeteners = survey[sweetener_columns]
sweeteners 
```

The dataframe now contains only the columns related to sweetener preferences. Just like with dairy, these columns represent whether a respondent uses a particular type of sweetener (1 for yes, 0 for no, and `NaN` if they didn't answer the sweetener questions).

Next, we'll rename these columns for better readability.

```python
sweetener_name_map = {
    'What kind of sugar or sweetener? (Granulated Sugar)': 'Granulated Sugar',
    'What kind of sugar or sweetener? (Artificial Sweetener)': 'Artificial Sweetener',
    'What kind of sugar or sweetener? (Honey)': 'Honey',
    'What kind of sugar or sweetener? (Maple Syrup)': 'Maple Syrup',
    'What kind of sugar or sweetener? (Stevia)': 'Stevia',
    'What kind of sugar or sweetener? (Agave Nectar)': 'Agave Nectar',
    'What kind of sugar or sweetener? (Brown Sugar)': 'Brown Sugar',
    'What kind of sugar or sweetener? (Raw Sugar)': 'Raw Sugar',
}
sweeteners = sweeteners.rename(columns=sweetener_name_map)
sweeteners
```

With the columns renamed, the dataframe is much easier to work with. Now, we need to handle the `NaN` values, which indicate that the survey question about sweeteners was not answered. Similar to our approach with dairy, we'll drop these rows, assuming respondents who didn't answer these questions also don't add sweeteners.

```python
sweeteners = sweeteners.dropna()
sweeteners
```

After dropping rows with `NaN` values, we are left with the responses from people who indicated their sweetener preferences. This cleaned dataset is now ready for analysis.

Now, let's calculate the percentage of people who use each type of sweetener, similar to what we did for dairy preferences.

```python
sweetener_preferences = sweeteners.mean() * 100
sweetener_preferences = sweetener_preferences.sort_values()
sweetener_preferences 
```

The series `sweetener_preferences` now shows the percentage of respondents who use each type of sweetener, sorted from least to most popular. Granulated sugar is by far the most popular, followed by honey and artificial sweetener. Just like with dairy, these percentages can add up to more than 100% because people can use multiple types of sweeteners.

Let's visualize these preferences using a horizontal bar chart to easily compare them.

```python

plt.figure(figsize=(8, 4))
plt.barh(sweetener_preferences.index, sweetener_preferences)
plt.xlabel("Percent")
plt.title("Sweetener Preferences")
plt.tight_layout()
plt.show()
```

This bar chart clearly illustrates the sweetener preferences. Granulated Sugar is significantly more popular than other options, indicating it's a staple for most coffee drinkers. Honey and Artificial Sweetener also have a notable following, while other options like Agave Nectar and Brown Sugar are less frequently chosen.

### Conclusion

This project aimed to analyze coffee enthusiast survey data to provide recommendations for a new plant-based coffee shop regarding dairy alternatives and sweetener options.

**Key Findings:**

*   **Dairy Preferences:** Among the plant-based dairy options, **Oat milk** was the most preferred (approximately 38.8% of respondents), followed by Almond milk (approximately 10.3%) and Soy milk (approximately 5.7%). Overall, Whole milk remains the most popular choice, but for a plant-based shop, Oat milk stands out as the clear leader.

*   **Sweetener Preferences:** **Granulated Sugar** is overwhelmingly the most popular sweetener (approximately 55.4%), followed by Honey (approximately 21.6%) and Artificial Sweetener (approximately 20.2%). Other sweeteners like Stevia, Maple Syrup, and Brown Sugar are less commonly used.

**Recommendations for the Plant-Based Coffee Shop:**

1.  **Dairy Alternatives:** The coffee shop should prioritize stocking **Oat milk** as its primary plant-based dairy alternative due to its high popularity. While Almond and Soy milk are less popular, offering them could cater to a broader range of preferences for those who don't prefer oat milk or have other dietary needs.

2.  **Sweeteners:** The coffee shop should definitely provide **Granulated Sugar** as the default sweetener. Additionally, offering **Honey** and an **Artificial Sweetener** would cover the preferences of a significant portion of customers. Other less popular sweeteners might be considered for a more extensive menu or as niche offerings.

**Overall Takeaway:**

To optimize customer satisfaction, the new plant-based coffee shop should focus its inventory on **Oat milk** as the leading dairy alternative and ensure a readily available supply of **Granulated Sugar**, **Honey**, and an **Artificial Sweetener**. These choices align with the most prominent preferences identified in the survey, allowing the shop to cater effectively to a large customer base while maintaining its plant-based focus.
